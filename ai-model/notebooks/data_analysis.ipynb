```python
from pathlib import Path
import matplotlib.pyplot as plt
from PIL import Image

RAW = Path('../dataset/raw')
classes = sorted(p for p in RAW.iterdir() if p.is_dir())
print(len(classes), 'classes')
```

```python
counts = {c.name: len(list(c.glob('*'))) for c in classes}
plt.figure(figsize=(10, 4))
plt.bar(counts.keys(), counts.values())
plt.xticks(rotation=90)
plt.ylabel('images')
plt.tight_layout()
```

```python
# Preview one image per class
fig, axes = plt.subplots(2, 5, figsize=(14, 6))
for ax, c in zip(axes.flat, classes):
    ax.imshow(Image.open(next(c.glob('*'))))
    ax.set_title(c.name, fontsize=7)
    ax.axis('off')
```
