```python
import sys
sys.path.insert(0, '..')
from training.train import load_datasets, build_model, compile_model
from pathlib import Path

train_ds, val_ds, classes = load_datasets(Path('../dataset/processed'), 32)
model, base = build_model(len(classes))
compile_model(model, 1e-3)
model.summary()
```

```python
history = model.fit(train_ds, validation_data=val_ds, epochs=5)
```

```python
import matplotlib.pyplot as plt
plt.plot(history.history['accuracy'], label='train')
plt.plot(history.history['val_accuracy'], label='val')
plt.legend(); plt.xlabel('epoch'); plt.ylabel('accuracy')
```
